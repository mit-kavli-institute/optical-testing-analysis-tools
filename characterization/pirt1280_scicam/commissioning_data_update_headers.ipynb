```python
import os
import sys
import astropy.io.fits as fits
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from glob import glob

```

```python
import os
from glob import glob
targetName = "WNTR25fejjz"
science_dir = os.path.join(os.path.expanduser("~"), "data", 
                           "images", "pirt_commissioning_data_pkg", 
                           "science",targetName)

science_image_paths = glob(os.path.join(science_dir, '*.fits'))

print(f"found {len(science_image_paths)} fits files in {science_dir}")
```

```python
# Import each file, update the header, save it back out

updated_RA = 9.880332
updated_DEC = 41.191133

for file in science_image_paths:
    print(f"working on {file}")
    with fits.open(file, mode='update') as hdul:
        hdr = hdul[0].header
        # Update headers here
        original_header = hdr.copy()
        hdr['RADEG'] = updated_RA
        hdr['DECDEG'] = updated_DEC
        hdr['OBJRA'] = updated_RA
        hdr['OBJDEC'] = updated_DEC
        
        print(f"Updated: RADEG: {original_header.get('RADEG')} -> {hdr.get('RADEG')}, DECDEG: {original_header.get('DECDEG')} -> {hdr.get('DECDEG')}")
        print(f"Updated: OBJRA: {original_header.get('OBJRA')} -> {hdr.get('OBJRA')}, OBJDEC: {original_header.get('OBJDEC')} -> {hdr.get('OBJDEC')}")

        # Number of images per dither
        print(f"hdr['N_IMGS_PER_DITHER']: {hdr.get('N_IMGS_PER_DITHER')}")
        n_imgs_per_dither = hdr.get('N_IMGS_PER_DITHER', "")
        hdr["DITHNIMG"] = n_imgs_per_dither
```
