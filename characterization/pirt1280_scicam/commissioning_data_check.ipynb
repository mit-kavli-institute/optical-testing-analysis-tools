```python
import os
import sys
import astropy.io.fits as fits
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from glob import glob

```

```python
BOX_PATH = os.path.join(os.getenv("HOME"), "Library", "CloudStorage", "Box-Box")

DATA_DIR = os.path.join(BOX_PATH, "WINTER", "WINTER – Public Data Workspace", "pirt_camera", 
                        "commissioning", "commissioning_data_pkg")

DARKS_DIR = os.path.join(DATA_DIR, "darks")
```

```python
# Darks

# load in all the darks, open them with fits, make a table with the: filename, filepath, exposure time, obstype
dark_files = glob(os.path.join(DARKS_DIR, "*.fits"))

print(f"Found {len(dark_files)} dark files.")

# set up a pandas dataframe to hold the information
df_darks = pd.DataFrame(columns=["filename", "filepath", "exposure_time", "obstype"])

for filepath in dark_files:
    with fits.open(filepath) as hdul:
        header = hdul[0].header
        data = hdul[0].data

        # Extract header information
        exposure_time = np.round(header.get("EXPTIME", np.nan), 4)
        obstype = header.get("OBSTYPE", "")
        
        # Make a dict to add to the pandas dataframe
        file_info = {
            "filename": os.path.basename(filepath),
            "filepath": filepath,
            "exposure_time": exposure_time,
            "obstype": obstype
        }

        # Append to dataframe
        df_darks = df_darks.append(file_info, ignore_index=True)
```
